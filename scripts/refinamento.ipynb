```python
import duckdb
```

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
df = con.execute("SELECT * FROM produtos").fetchdf()
df.head(10)
```

```python
df2 = df.drop(columns=['id_categoria', 'id_fornecedor'])
df2.head(10)
```

```python
con.execute("""
CREATE TABLE IF NOT EXISTS dim_produtos (
    id_produto BIGINT,
    nm_produto TEXT,
    vl_preco FLOAT
    )
""")
```

```python
con.execute("INSERT INTO dim_produtos SELECT * FROM df2")
```

```python
dim_produtos = con.execute("SELECT * FROM dim_produtos").fetchdf()
dim_produtos.head(10)
```

```python
con.execute("SHOW TABLES").fetchdf()
```

```python
con.close()
```
